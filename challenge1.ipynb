```python
import numpy as np
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit
from qiskit.transpiler import generate_preset_pass_manager
from qiskit.visualization import plot_histogram

from qiskit_ibm_runtime import (
    QiskitRuntimeService,
    SamplerV2 as Sampler
)

# IBM Quantum connection
service = QiskitRuntimeService(
    channel="ibm_quantum_platform",
    token="W0MxmzJ3rHSJhCNbzSUbtYiQ83XXwWtn3q5P-37DdNRU",
    instance="crn:v1:bluemix:public:quantum-computing:us-east:a/704faf63a9da4de092a44a4307e4508c:42ab98ac-e030-425a-9c6b-d4c631b68a65::"
)

# Select an available real quantum computer
backend = service.least_busy(
    operational=True,
    simulator=False,
    min_num_qubits=2
)

print("Backend:", backend.name)
```

```python
# Challenge 1 - Circuit A
qc_A = QuantumCircuit(1, 1)

qc_A.x(0)
qc_A.measure(0, 0)

print("Circuit A:")
print(qc_A.draw())
```

```python
pm = generate_preset_pass_manager(
    target=backend.target,
    optimization_level=3
)

qc_A_isa = pm.run(qc_A)

sampler = Sampler(mode=backend)

job_A = sampler.run([qc_A_isa], shots=100)
result_A = job_A.result()

counts_A = result_A[0].data.c.get_counts()

print("Circuit A results:", counts_A)

plot_histogram(counts_A)
plt.show()
```
